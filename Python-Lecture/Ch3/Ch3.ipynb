#List


*   list --> mutable
*   use an index to access each element/item in the list



```python
mid_scores = [100, 98, 85, 79, 90] #midterm scores



#you can use an index to access the element in the list
print(mid_scores[0])
print(mid_scores[4])
print(mid_scores[-1]) #-1 means --> last index
```

```python
mid_scores.append(81) #append --> add the new element to the tail of the list
mid_scores
```

```python
# list.pop(i): Removes the element at index i from list
mid_scores.pop(1 )
mid_scores
```

```python
#list.remove(v): Removes the first element whose value matches v
mid_scores.remove(81)
```

```python
fruits = ["apple", "banana", "blueberry"] #fruit

print(fruits[1])
```

# Dictionary


*   Mutable
*   key:value pair
*   pop to remove the item in the Dictionary



```python
fruit_price = {"apple":2.56, "banana":2.10, "blueberry":4.10}

#to access the price of apple
print(fruit_price["apple"]) #you can access the value via the corresponding key
```

```python
#To remove --> pop
fruit_price.pop("blueberry")

fruit_price
```
